## Pandas Basics

```python
import pandas as pd
s = pd.Series([10,20,30,40,50], name = 'numbers')
print(s)
```

```python
print("Date type",s.dtype)
print("Value: ",s.values)
print("Index: ",s.index)
print("Name: ",s.name)
```

## Indexing

```python
print(s[0])
s[0]
```

```python
print(s[0:3])
s[0:4]
```

## Location Based Indexing

```python
print(s.iloc[3])
```

```python
print(s.iloc[[3,4,1]])
```

```python
s.name = "Calories"
print(s)
```

```python
index = ['Apple', 'Banana', 'Orange', 'Guava', 'Mango']
s.index = index
print(s['Mango'])
```

```python
print(s)
```

```python
a = pd.Series([1,2,3,4,5])
a.name = "Roll no."
a.index = ['karan', 'pratik', 'prince','lalit', 'Gopal']
print(a)
```

## Label Based Indexing

```python
print(s.loc['Mango'])
print(a.loc['karan'])
print(s.loc[['Mango', 'Apple']])

print(s['Apple' : 'Guava'])
```
