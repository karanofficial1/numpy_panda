```python
counts = [577, 314]
labels= ['male', 'female']
plt.figure(figsize= (6,6))
plt.pie(counts, labels = labels, autopct ='%1.1f%%', startangle = 140, colors= ['red', 'pink'], explode= (0,0.1), shadow= True)
plt.title("Male Female Count Pie Chart")
plt.show()
```

```python
data = {
    'salary' : [55000, 30000, 35000, 60000 ,45000 ]
}
df = pd.DataFrame(data)
df.head()
```

```python
plt.hist(df['salary'], bins= 5 , color='skyblue')
plt.show()
```

```python
df.drop(index=5, inplace= True)
```

```python
df['Dept'] = ["HR", "Finance", "IT", "Finance", "IT"]
df.head()
df.info()
```

## Bivariate- Numerical -Numerical

```python
df['Age']= [25,35,26,20,27]
```

```python
df.head()
```

## Scatter plot

```python
plt.scatter(df['Age'], df['salary'], color= 'red')
plt.show()
```

## Line Plot

```python
plt.plot(df["Age"], df["salary"], color="red", marker = "*", linestyle="-", linewidth= "2")
plt.grid()
plt.show()
```

```python
sort_age = df.sort_values("Age")
plt.plot(sort_age["Age"], df["salary"])
plt.grid()
plt.show()
```

## Bar Chart

```python
plt.bar(sort_age["Age"], df["salary"], color = "green", align= "edge")
plt.show()
```

```python
plt.bar(sort_age["Age"], df["salary"], color = "green", align= "center")
plt.show()
```

## Bivariate Numerical/Categorical

```python
hr_sal = df[df["Dept"] == "HR"]["salary"]
it_sal = df[df["Dept"] == "IT"]["salary"]
finance_sal = df[df["Dept"] == "Finance"]["salary"]
```

## Boxplot

```python
plt.boxplot([hr_sal, it_sal, finance_sal], label= ['HR', "IT", "Finance"])
plt.legend()
plt.show()
```

## Pie Chart

```python
salary_by_dept = df.groupby("Dept")["salary"].sum()
plt.pie(salary_by_dept, labels = salary_by_dept.index, shadow=True, explode=[0,0.1,0.1], autopct="%1.2f%%")
plt.axis("equal")
print(salary_by_dept)
```

## Bar Plot

```python
hr_mean = sum(hr_sal)/len(hr_sal)
it_mean = sum(it_sal)/len(it_sal)
finance_mean = sum(finance_sal)/len(finance_sal)
plt.bar(["HR", "IT", "Finance"], [hr_mean, it_mean, finance_mean])
plt.grid()
plt.show()
```

## Multivariate Analysis: 3 numerical columns

```python
df["Experience"] = [5,3,4,8,6]
```

```python
df.head()
```

## Bubble Plot

```python
plt.scatter(df["Age"], df["salary"], s=df["Experience"]*50, color= "orange", edgecolors="black")
plt.title("Age vs salary vs experience")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.grid()
plt.show()
```

## 2 Numerical and 1 Categorical column

```python
# plt.scatter(df["Age"], df["salary"],)
# plt.title("Age vs salary vs Dept")
# plt.xlabel("Age")
# plt.ylabel("Salary")
# plt.show()
color= {"HR":"Yellow", "IT":"blue", "Finance":"Orange"}
for dept, color in color.items():
    df_dept = df[df["Dept"] == dept]
    plt.scatter(df_dept["Age"], df_dept["salary"], c= color, label= dept)
plt.title("Age vs Salary by Department")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.legend()
plt.show()
```

# OOP API

```python
fig, axs = plt.subplots(1,3, figsize= (20,15))
```

## Line Plot
